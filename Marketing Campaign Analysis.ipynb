```python
import pandas as pd
```

```python
file = "Marketing Campaign Performance (Raw Dataset).xlsx"

campaigns = pd.read_excel(file, sheet_name = 'Campaigns')
campaigns.head()
```

```python
# Identifying duplicate values

campaigns.duplicated().sum()
```

```python
# Droping Duplicate Values
campaigns = campaigns.drop_duplicates()
campaigns
```

```python
campaigns.duplicated().sum()
```

```python
# Identifying missing values

campaigns.isnull().sum()
```

```python
# Handiling missing values

campaigns['Objective'] = campaigns['Objective'].fillna(campaigns['Objective'].mode()[0])

campaigns['Budget'] = campaigns['Budget'].fillna(campaigns['Budget'].median())

campaigns['Target_Audience'] = campaigns['Target_Audience'].fillna(campaigns['Target_Audience'].mode()[0])

campaigns
```

```python
campaigns.isnull().sum()
```

```python
# Converting Dates into DateTime

campaigns['Start_Date'] = pd.to_datetime(campaigns['Start_Date'])

campaigns['End_Date'] = pd.to_datetime(campaigns['End_Date'])
campaigns
```

```python
customers = pd.read_excel(file, sheet_name = 'Customers')
customers
```

```python
# Identifying Duplicate Records

customers.duplicated().sum()
```

```python
# Droping Duplicate Values
customers = customers.drop_duplicates()
customers
```

```python
# Identifying missing values

customers.isnull().sum()
```

```python
# Handiling missing values

customers['Gender'] = customers['Gender'].fillna(customers['Gender'].mode()[0])

customers['City'] = customers['City'].fillna(customers['City'].mode()[0])

customers['Customer_Segment'] = customers['Customer_Segment'].fillna(customers['Customer_Segment'].mode()[0])
customers
```

```python
# Converting Dates into DateTime
customers['Signup_Date'] = pd.to_datetime(customers['Signup_Date'])
customers
```

```python
performance = pd.read_excel(file, sheet_name = 'Campaign_Performance')
performance
```

```python
# Identifying Duplicate Values

performance.duplicated().sum()
```

```python
# Removing Duplicate Values

performance = performance.drop_duplicates()
performance
```

```python
performance.duplicated().sum()
```

```python
# Identifying missing values

performance.isnull().sum()
```

```python
# Handiling missing values

performance['Clicks'] = performance['Clicks'].fillna(performance['Clicks'].median())

performance['Spend'] = performance['Spend'].fillna(performance['Spend'].median())

performance['Revenue'] = performance['Revenue'].fillna(performance['Revenue'].median())

performance['Device'] = performance['Device'].fillna(performance['Device'].mode()[0])

performance['Audience_Type'] = performance['Audience_Type'].fillna(performance['Audience_Type'].mode()[0])

performance
```

```python
# Converting Date

performance['Date'] = pd.to_datetime(performance['Date']).dt.date
performance
```

```python
# Saving files into CSV

campaigns.to_csv("cleaned_campaigns.csv", index = False)
customers.to_csv("cleaned_customers.csv", index = False)
performance.to_csv("cleaned_performance.csv", index = False)
```
